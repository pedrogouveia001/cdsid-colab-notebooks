```python
import math
A = float(input())
B = float(input())
L = float(input())

def f(x):
  x = math.exp(-x**2) - math.cos(x)
  return x

if f(A)*f(B) < 0:
  if f(A) > f(B):
    Xo = A
    A = B
    B = Xo
  Xo = (A*f(B) - B*f(A))/(f(B) - f(A))
else:
  print("não há raiz neste intervalo")

while L < abs(f(Xo)):
  if f(B)*f(Xo) <= 0:
    Xo = (Xo * f(B) - B * f(Xo))/(f(B) - f(Xo))
  elif f(A)*f(Xo) < 0:
    Xo = (A * f(Xo) - Xo * f(A))/(f(Xo) - f(A))

print("%s\n%s" % (str(round(Xo, 7)), str(round(f(B) - f(A), 7))))
```
