```python
import math

def fatoracao(a):
  if a != 0:
    result = a*fatoracao(a-1)
  else:
    result = 1 
  return result

xp = float(input())

y = [[11.02, 13.46, 16.44, 20.08, 24.53, 29.96, 36.59, 44.70], [], [], [], [], [], [], []]
Polinomio = [1, 1, 1, 1, 1, 1, 1, 1]
xinicial = 2.4
dist = 0.2
i = 1
j = 0
while (i < 8):
  for j in range(0, 8-i):
    y[i].append(y[i-1][j+1] - y[i-1][j])
  i+=1

i = 0
j = 0
while (i < 8):
  for j in range (0, 8):
    if j == i:
      auxj = fatoracao(j)
      Polinomio[i] *= (y[j][0])/(auxj*(dist**j))
    if j < i:
      Polinomio[i] *= (xp - (xinicial + dist*j))
  i+=1
Resultado = int((10**7)*abs(math.fsum(Polinomio) - math.exp(xp)))  - int((10**2)*abs(y[7][0]))
print(Resultado)
```
