```python
import math
n = int(input())
x = []
y = []
g = []
h = float()
Trapezio = float()
Simpson = float()

i = 0
while i < n:
  a = float(input())
  x.append(a)
  i+=1

h = x[1] - x[0]

i=0
while i < n:
  a = float(input())
  y.append(a)
  g.append(a*x[i])
  i+=1

Trapezio = ( g[n-1] + g[0])/2
Simpson = (g[n-1] + g[0])

i=0
while i < n-1:
  if i % 2 == 0:
    Simpson += 2*g[i]
  if i % 2 == 1:
    Simpson += 4*g[i]
  Trapezio += g[i]
  i+=1

Trapezio = h*Trapezio
Simpson *= h/3

print(abs(int((round(Simpson, 3))*1000 - int((round(Trapezio, 3))*1000))))
```
