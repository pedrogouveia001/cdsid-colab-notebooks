```python
def funcao(a, b, c):
  return a*c**3 - b*c + 1

Trapezio = float(0)
n = int(1)

xi = float(input())
xf = float(input())
a = float(input())
b = float(input())
k = int(input())
Delta = float(input())

h = (xf - xi)/(k-1)
Simpson = (funcao(a, b, xi) + funcao(a, b, xf))

i =1
while i < k-1:
  if i % 2 == 0:
    Simpson += funcao(a, b, xi + i*h)*2
  if i % 2 == 1:
    Simpson += funcao(a, b, xi + i*h)*4
  i+=1
Simpson *= h/3

while abs(Trapezio - Simpson) > Delta:
  n += 1
  Trapezio = (funcao(a, b, xf) + funcao(a, b, xi))/2
  i=1
  while i < n-1:
    Trapezio += funcao(a, b, xi + i*((xf - xi)/(n-1)))
    i+=1
  Trapezio *= ((xf - xi)/(n-1))
  
print(n)
```
