```python
grafo = {
  'A' : ['B','C', 'D'],
  'B' : ['A','C', 'D', 'E'],
  'C' : ['A' ,'B', 'E'],
  'D' : ['A', 'B', 'E', 'F'],
  'E' : ['B', 'C', 'D', 'F'],
  'F' : ['D','E']
}

visitados = [] 
fila = []   

def bfs(visitados, grafo, no): 
  visitados.append(no)
  fila.append(no)

  while fila:          
    numnos = fila.pop(0) 
    print (numnos, end = " ") 

    for i in grafo[numnos]:
      if i not in visitados:
        visitados.append(i)
        fila.append(i)

def dfs(visitados, grafo, no):
    if no not in visitados:
      print(no, end = " ")
      visitados.append(no)
      for i in grafo[no]:
        dfs(visitados, grafo, i)


a = int(0)
while(a == 0):
  print("Digite o nó inicial: " )
  noinicial = str(input())
  if ((noinicial == 'A')or(noinicial == 'B')or(noinicial.upper() == 'C')or(noinicial == 'D')or(noinicial == 'E')or(noinicial == 'F')):
    print("A árvore gerada pelo BFS é dada por:")
    bfs(visitados, grafo, noinicial)
    print("")
    visitados = []
    print("A árvore gerada pelo DFS é dada por:")
    dfs(visitados, grafo, noinicial)    
    a=1
  else:
    print("Esse nó não existe, por favor escolha um Nó entre os seguintes: A,B,C,D,E,F")
    print('')
```
