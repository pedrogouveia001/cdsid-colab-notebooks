```python
import math
L = float(input())
H = float(input())
n = int(input())
dif_max = float(0)
i = int(0)
j = int(0)

def soma (a,b):
  i = 0
  auxSoma = float(0)
  while i < b:
    auxSoma += a/b
    i+=1
  return auxSoma

def modifica (k):
  if k < 0:
    k = -k
  return k

j = n-50
while j < n+50:
  if dif_max <= modifica((soma(H*L, j) - H*L)) :
    dif_max = modifica((soma(H*L, j) - H*L))
    n_max = int(j)
  j+=1

print("n_max = {0:d}\nerro_maximo = {1:.14f}".format(n_max, dif_max))
```
