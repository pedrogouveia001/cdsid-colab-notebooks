```python
import numpy as opnp
import math as opmat
X = []
FX = []
auxa = [0., 0., 0, 0., 0.]
auxb = [0., 0., 0., 0., 0.]
X.append([0., 0., 0, .0])
X.append([0., 0., 0, .0]) 
X.append([0., 0., 0, .0]) 
X.append([0., 0., 0, .0])  
FX = [0., 0., 0, .0]

def MMQ(a1,a2,a3,a4,t):
  x1 = a1*t*t
  x2 =  a2*t
  x3 = a3*opmat.cos(t) 
  x4 = a4
  return (x1+x2+x3+x4)
  
i = 0
j= 0

while i < 5:
  auxa[i] = float(input())
  X[0][0] = 1 + X[0][0]
  X[1][1] =  opmat.cos(auxa[i])*opmat.cos(auxa[i]) + X[1][1]
  X[2][2] = auxa[i]*auxa[i] + X[2][2]
  X[3][3] = opnp.power(auxa[i], 2)*opnp.power(auxa[i], 2) + X[3][3]

  X[0][1] = opmat.cos(auxa[i]) + X[0][1]
  X[0][2] = auxa[i] + X[0][2]
  X[0][3] = opnp.power(auxa[i], 2) + X[0][3]
  X[1][2] =  opmat.cos(auxa[i])*auxa[i] + X[1][2]
  X[1][3] =  opmat.cos(auxa[i])*opnp.power(auxa[i], 2) + X[1][3]
  X[2][3] = auxa[i]*opnp.power(auxa[i], 2) + X[2][3]

  X[1][0] = opmat.cos(auxa[i]) + X[1][0]
  X[2][0] = auxa[i] + X[2][0]
  X[3][0] = opnp.power(auxa[i], 2) + X[3][0]
  X[2][1] =  opmat.cos(auxa[i])*auxa[i] + X[2][1]
  X[3][1] =  opmat.cos(auxa[i])*opnp.power(auxa[i], 2) + X[3][1]
  X[3][2] = auxa[i]*opnp.power(auxa[i], 2) + X[3][2]
  i+=1

while j < 5:
  auxb[j] = float(input())
  FX[0] = auxb[j] + FX[0]
  FX[1] = auxb[j]*opmat.cos(auxa[j]) + FX[1]
  FX[2] = auxb[j]*auxa[j] + FX[2]
  FX[3] = auxb[j]*opnp.power(auxa[j], 2) + FX[3]
  j+=1


   

[a1, a2, a3, a4] = opnp.linalg.solve(X, FX)

alinhamento1 = round(MMQ(a4,a3,a2,a1,1.1),6)
alinhamento2 = round(MMQ(a4,a3,a2,a1,4.3),6)

print(alinhamento1,alinhamento2)
```
