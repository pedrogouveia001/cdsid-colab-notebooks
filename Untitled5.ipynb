```python
def pivotamento(A):
    """A função recebe uma matriz A (n x n) e
    retorna a matriz zerada na diagonal inferior
    esquerda."""
    for i in range(len(A) - 1):
        for j in range(i + 1,len(A)):
            n1 = A[j][i]
            if n1 == 0:
                for k in range(i,len(A)):
                    if A[k][i] != 0:
                        swap(A, i, k)
                        n1 = A[j][i]
            n2 = A[i][i]
            if n2 == 0:
                continue
            div = n1 / n2
            A[j] = sumVector(multVector(A[i], -div), A[j])
    return A


def multVector(v, x):
    newV = []
    for i in range(len(v)):
        newV += [v[i] * x]
    return newV

def sumVector(v1, v2):
    newV = []
    for i in range(len(v1)):
        newV += [v1[i] + v2[i]]
    return newV

def swap(A, i, j):
    newI = []
    newJ = []
    for index in range(len(A) + 1):
        newI += [A[j][index]]
        newJ += [A[i][index]]
    A[i] = newI
    A[j] = newJ

def printMatrix(A):
    for i in A:
        print ("|\t"),
        for j in i:
            print ("%.1f" %(j), "\t"),
        print ("\t|")

A = [[6 ,1, 0, -7], [8, 6, 1, 15], [0, 8, 6, 14]]
print(pivotamento(A))
```
