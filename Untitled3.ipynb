```python
def pivotamento(A):
    """A função recebe uma matriz A (n x n) e
    retorna a matriz zerada na diagonal inferior
    esquerda."""
    for i in range(len(A) - 1):
        for j in range(i + 1,len(A)):
            n1 = A[j][i]
            if n1 == 0:
                for k in range(i,len(A)):
                    if A[k][i] != 0:
                        substituicao(A, i, k)
                        n1 = A[j][i]
            n2 = A[i][i]
            if n2 == 0:
                continue
            div = n1 / n2
            A[j] = soma(multiplicacao(A[i], -div), A[j])
    return A


def multiplicacao(v, x):
    a = []
    for i in range(len(v)):
        a += [v[i] * x]
    return a

def soma (v1, v2):
    b = []
    for i in range(len(v1)):
        b += [v1[i] + v2[i]]
    return b

def substituicao(A, i, j):
    a = []
    b = []
    for k in range(len(A)):
        a += [A[j][k]]
        b += [A[i][k]]
    A[i] = a
    A[j] = b

A = [[6, 1, 0], [8, 6, 1], [0, 8, 6]]
print(A)
print(pivotamento(A))
```
