```python
N = int(input())
a = int()
lst = []
lst_ = []
M = []
M_ = []
b = []
c = []
x = []
z = []
sum = float(0.)
erro_sem = float()
erro_com = float()
def mod(l):
  if l < 0:
    l = -l
  return l
for i in range(0, N):
  for j in range(0, N):
    lst.append(0.)
    lst_.append(0.)
  M.append(lst)
  M_.append(lst_)
  lst = []
  lst_ = []
  if i == 0:
    b.append(7.)
    c.append(7.)
  elif i < N-1:
    b.append(15.)
    c.append(15.)
  else:
    b.append(14.)
    c.append(14.)
  x.append(0.)
  z.append(0.)
for i in range(0, N):
  for j in range(0, N):
    if i == j:
      M[i][j] = 6
      M_[i][j] = 6
    elif i == j-1:
      M[i][j] = 1
      M_[i][j] = 1
    elif i == j+1:
      M[i][j] = 8
      M_[i][j] = 8
    else:
      M[i][j] = 0
      M_[i][j] = 0
for j in range(0, N):
  for i in range(0, N):
    if j < i:
      if M[i][j] != 0:
        r = M[i][j] / M[j][j]
        for k in range(0, N):
          M[i][k] = M[i][k] - r*M[j][k]
        b[i] = b[i] - r * b[j]
    if M_[i][j] != 0:
      if M_[i][j] > M_[j][j]:
        for k in range(0, N):
          M_[j][k], M_[i][k] = M_[i][k], M_[j][k]
        c[j], c[i] = c[i], c[j]
      r_ = M_[i][j]/M_[j][j]
      for k in range(0, N):
        M_[i][k] = M_[i][k] - r_*M_[j][k]
      c[i] = c[i] - r_*c[j]
j = N-1
while j >= 0:
  x[j] = b[j]/M[j][j]
  z[j] = c[j]/M_[j][j]
  i = N-1
  while i > j:
    x[j] += (-M[j][i])*(x[i])/(M[j][j])
    z[j] += (-M_[j][i])*(z[i])/(M_[j][j])
    i -= 1
  j -= 1
for i in range(0, N):
  sum += mod(x[i])
erro_sem = mod(N - sum)
sum = 0
for i in range(0, N):
  sum += mod(z[i])
erro_com += mod(N - sum)
print("erro_sem_pivo = ", round(erro_sem,6), "\nerro_com_pivo = ", round(erro_com,6))
```
