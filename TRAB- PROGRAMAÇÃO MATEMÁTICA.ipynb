# Questão 1

```python
import numpy as np
from scipy.optimize import minimize

#função objetivo
def objective_function(x):
    return 5 * np.exp(2 * x[0]) + 2 * np.exp(3 * x[1])

#define as restrições
constraints = [{'type': 'ineq', 'fun': lambda x: x[0]},
               {'type': 'ineq', 'fun': lambda x: x[1]},
               {'type': 'eq', 'fun': lambda variables: variables[0] + variables[1] - 100}]

#chute do inicial para as variáveis
initial_guess = [0, 0]

#result = minimize(objective_function, initial_guess, constraints=constraints, method='SLSQP')
result = minimize(objective_function, initial_guess, constraints=constraints, method='trust-constr')

#verifica o resultado
if result.success:
    print("Status da otimização:", result.message)
    print("Valor da função objetivo:", result.fun)
    print("Valor de x:", result.x[0])
    print("Valor de y:", result.x[1])
else:
    print("A otimização não foi bem-sucedida:", result.message)
```

```python
components = {
    1: {'R': [0.6, 0.8, 0.9], 'C': [1000, 2000, 3000]},
    2: {'R': [0.7, 0.8, 0.9], 'C': [3000, 5000, 6000]},
    3: {'R': [0.5, 0.7, 0.9], 'C': [2000, 4000, 5000]}
}

valid_combinations = []

for combo1 in components[1]['C']:
    for combo2 in components[2]['C']:
        for combo3 in components[3]['C']:
            total_cost = combo1 + combo2 + combo3
            if total_cost <= 10000:
                selected_R = [
                    components[1]['R'][components[1]['C'].index(combo1)],
                    components[2]['R'][components[2]['C'].index(combo2)],
                    components[3]['R'][components[3]['C'].index(combo3)]
                ]
                # print(selected_R, combo1,combo2,combo3)
                product_R = selected_R[0] * selected_R[1] * selected_R[2]
                valid_combinations.append((combo1, combo2, combo3, product_R))

print("Combinações válidas (custo total < 10000):")
for combo in valid_combinations:
    print("Combinação:", combo[:3], "Custo Total:", sum(combo[:3]), "Produto R:", combo[3])
```

```python
components = {
    1: {'R': [0.6, 0.8, 0.9], 'C': [1000, 2000, 3000]},
    2: {'R': [0.7, 0.8, 0.9], 'C': [3000, 5000, 6000]},
    3: {'R': [0.5, 0.7, 0.9], 'C': [2000, 4000, 5000]}
}

valid_combinations = []
best_combination = None
best_product_R = 0

for combo1 in range(3):
    for combo2 in range(3):
        for combo3 in range(3):
            total_cost = components[1]['C'][combo1] + components[2]['C'][combo2] + components[3]['C'][combo3]
            if total_cost <= 10000:
                selected_R = [
                    components[1]['R'][combo1],
                    components[2]['R'][combo2],
                    components[3]['R'][combo3]
                ]
                product_R = selected_R[0] * selected_R[1] * selected_R[2]
                valid_combinations.append((combo1, combo2, combo3, product_R))

                if product_R > best_product_R:
                    best_product_R = product_R
                    best_combination = (combo1, combo2, combo3, product_R)

print("Todas as combinações válidas (custo total < 10000)")
for combo in valid_combinations:
    print("Combinação:", combo[:3])
    print("Custo Total:", sum([components[1]['C'][combo[0]], components[2]['C'][combo[1]], components[3]['C'][combo[2]]]), "R:", combo[3])


print("\nMelhor Combinação:")
print("Combinação:", best_combination[:3])
print("Custo Total:", sum([components[1]['C'][best_combination[0]], components[2]['C'][best_combination[1]], components[3]['C'][best_combination[2]]]),"R:", best_combination[3])
```
