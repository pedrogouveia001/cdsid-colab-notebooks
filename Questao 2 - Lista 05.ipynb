```python
import math

x = []
y = [[], [], [], []]
Polinomio = [1, 1, 1, 1]

AuxA = float()
a = 0
b = 0

while (a < 8):
  AuxA = float(input())
  if (a < 4):
    x.append(AuxA)
  else:
    y[0].append(AuxA)
  a+=1


xp = float(input())

c=0
while (c < 3):
  y[1].append((y[0][c+1] - y[0][c])/(x[c+1] - x[c]))
  c+=1
  
c = 0
while c < 2:
    y[2].append((y[1][c+1] - y[1][c])/(x[c+2] - x[c]))
    c+=1


y[3].append((y[2][1] - y[2][0])/(x[3] - x[0]))

for a in range (0, 4):
  for b in range (0, 4):
    if b == a:
      Polinomio[a] *= y[b][0]
    if b < a:
      Polinomio[a] *= (xp - x[b])

print(int(abs(round(10000*y[3][0], 2))))
print(int(round(math.fsum(Polinomio),2)))
```
